```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("train.csv")
```

```python
#bar chart survived by class

fig,ax=plt.subplots()

classSurvivedRate=df.groupby("Pclass")["Survived"].sum()

ax.bar(classSurvivedRate.index,classSurvivedRate.values,color=['purple','violet','pink'])
ax.set_xticks([1,2,3])

ax.set_title('Class Survied Rate')
ax.set_xlabel('PClass')
ax.set_ylabel('Survived')
```

```python
#hbar

fig,ax=plt.subplots()

ax.barh(classSurvivedRate.index,classSurvivedRate.values, color=['pink','green','red'])
ax.set_yticks([1,2,3])

ax.set_title("abc")

```

```python
#histogram

fig,ax=plt.subplots()

ax.hist(df['Age'],edgecolor='black')

ax.set_title("abc")
```

```python
#SurvivedByAge

fig,ax=plt.subplots()

ax.hist(df[df['Survived']==1]['Age'],bins=20,edgecolor='black')
```

```python
#survived male and female histogram

fig,ax=plt.subplots(1,2) #1,2 means total two hist

ax[0].hist( df[(df["Survived"]==1) & (df["Sex"]=='male')]['Age'] , edgecolor='black', color='blue' , bins=20 )
ax[0].set_title("Male age survived")
ax[0].set_xlabel('Age')
ax[0].set_ylabel('Freq')

ax[1].hist( df[(df['Survived']==1) & (df['Sex']=='female')]['Age'] , edgecolor='black', color='pink' , bins=20)
ax[1].set_title("Female age survived")
ax[1].set_xlabel('Age')
ax[1].set_ylabel('Freq')
```
