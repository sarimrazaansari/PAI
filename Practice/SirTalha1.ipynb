```python
import pandas as pd
```

```python
data={
    "Title":["Harry Potter", "One Piece", "Attack on Titan","Blue Box"],
    "Author":["JK Rowling","Oda","Hajime Isayama","Kouji Miura"],
    "Price":[1200,1500,2200,1800]
}

df=pd.DataFrame(data)
df
```

```python
genre=pd.Series(["Fantasy","Adventure","Action","Romance"]) # can only add series in df
df["Genre"]=genre # adding Genre at end
#adding more
stock=pd.Series([15,25,10,30])
df["Stock"]=stock
df
```

```python
#remove particular cols
df.drop(columns="Genre",inplace=True)  #change in current df
df
```

```python
#insert at Particular position
#df.insert(index to add , name of col , data)

df.insert(2,"Genre",genre)
df
```

```python
def checkStock(bookTitle):
    bookData=df[df['Title']==bookTitle]

    if not bookData.empty:
        stockCount=bookData["Stock"].iloc[0]
        print(f"Book Name: {bookTitle} | Stock: {stockCount}")
    else:
        print("Book Not Found")

checkStock("Harry Potter")
checkStock("Naruto")
```

```python
totalInventoryValue=(df["Price"] * df["Stock"]).sum()
print("Total Value in invertory: ",totalInventoryValue)
```
